#### <strong>Part 1: Fibonacci With Test Driven Development</strong>

```python
# To Do: Write fibonacci function both iterative and recursive to show importance of Test Driven Development. 
# Motivation: oftentimes, you need to cross check the results of a function with another similar output for QA and data 
# quality purposes (completeness and accuracy from DQ dimensions). 

# Fibonacci Sequence: The Fibonacci sequence is a series of numbers in which each number is the sum of the two preceding ones, usually starting with 0 and 1. 
# Formula: fib(n) = fib(n-1) + fib(n-2) with base cases fib(0) = 0, fib(1) = 1. 
# @param: n: the position in the Fibonacci sequence to calculate.
# @return: the nth Fibonacci number. 
def fibonacci_recursive(n):
    # Sanity check: cannot have n less than 0. 
    if n <= 0: 
        print("Error: n must be a positive integer.")
        return 0
    if n <= 2:
        return 1 
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2) 
    
# @param: n: the position in the Fibonacci sequence to calculate iteratively.
# @return: the nth Fibonacci number. 
def fibonacci_iterative(n):
    if n <= 0:
        print("Error: n must be a positive integer.")
        return 0
    num1 = 0
    num2 = 1
    # _ means in Python that the variable is a throwaway variable, meaning that it is not going to be used in the code. 
    for _ in range(n):
        num1 = num2 
        num2 = num1 + num2 
    return num1 
```

```python
# Write unit tests for both the recursive and iterative Fibonacci functions to ensure they produce the same results for a range of input values. 
class TestFibonacciFunctions(ut.TestCase): 
    # Test 1: Base case: n == 2, should return 1 for both functions. : tech
    def test_fibonacci_function_base_case(self): 
        # assertEquals: determines if the values tested are equal. 
        test1 = fibonacci_recursive(2)
        test2 = fibonacci_iterative(2)
        print("Testing base case: \n")
        print(self.assertEqual(test1, test2)) 
        
    # Test 2: General case: n == 5, should return 5 for both functions. 
    def test_fibonacci_function_general_case(self):
        test1 = fibonacci_recursive(5)
        test2 = fibonacci_iterative(5)
        print("Testing equality: \n")
        print(self.assertEqual(test1, test2))
        
    # Test 3: Check for unusual situations, such as negative numbers, and ensure that the functions handle them gracefully. 
    def test_fibonacci_function_negative_case(self):
        # Test 3: negative case: n == -1, should return 0 for both functions. 
        test1 = fibonacci_recursive(-1)
        test2 = fibonacci_iterative(-1)
        print("Testing negative case: \n")
        print(self.assertEqual(test1, test2))
        
    # To Do: Write a function that takes a list of numbers and then tests the Fibonacci functions for each number in the list.  
    def test_fibonacci_function_list_of_numbers(self):
        lst = [1,2,3,4,5]
        lst2 = [-5,-4,-3,-2,-1]
        for num in lst:
            # Check that both Fibonacci functions are giving same result (sanity check)
            fib_test1 = fibonacci_recursive(num) 
            fib_test2 = fibonacci_iterative(num)
            print("Test the Fibonacci list of numbers: ")
        print(self.assertEqual(fib_test1, fib_test2))
```

#### Object oriented programming: (OOP) is a programming paradigm that uses objects and classes to organize code. It allows for encapsulation, inheritance, and polymorphism, making it easier to manage and maintain complex software systems. 
#### init: The __init__ method is a special method in Python classes that is automatically called when a new instance of the class is created. It is used to initialize the object's attributes and set up any necessary state. 

#### Source for basic Python object oriented programming: https://www.w3schools.com/python/python_oop.asp

```python
# Class: Starbucks represents a Starbucks store with its name, location, and menu. 
class Starbucks: 
    def __init__(self, name, location, menu):
        self.name = name 
        self.location = location 
        self.menu = menu 

    def display_info(self):
        # Self is a Python way of referring to the instance of the class. It allows you to access the attributes and methods of the class within its own methods. 
        print(f"Starbucks Name: {self.name}")
        print(f"Location: {self.location}")
        print("Menu:")
        for item in self.menu:
            print("Item ordered in Starbucks", item) 
```

```python
# Barista: Barista represents the barista working at a Starbucks store. It has a name and a method to prepare drinks. 
class Barista: 
    def __init__(self, name, order):
        self.name = name 
        self.order = order 

    def prepare_drink(self, drink):
        print("Drink prepared in Starbucks", drink) 
```

```python
# Customer: you :) The customer is the person who orders drinks at Starbucks. It has a name and a method to place an order. 
class Customer: 
    def __init__(self, name, drink):
        self.name = name 
        self.drink = drink

    def place_order(self, drink):
        print("I would like to place an order for a, ", str(drink))
```

```python
#### Python example of determining what type of drink to order given the drinks and the prices 
drinks = {
    "Latte": 4.50,
    "Cappuccino": 4.00,
    "Espresso": 3.50,
    "Pink Drink": 6.70, 
    "Mocha": 5.00 
}

# Barista: Barista represents the barista working at a Starbucks store. It has a name and a method to prepare drinks. 
barista = Barista("Elizabeth", "Latte")
print("Barista", barista.name, "is preparing a", barista.order, ".")

customers = {
    "Jason": "Hot Chocolate", #Jason is a Senior Analyst at KPH and loves hot chocolate. 
    "Cole": "Latte", #Cole is a Senior Analyst at KPH and loves lattes. 
    "John": "Mocha", #John is a Service Delivery Technician at KPH and loves mochas.
    "Marco": "Cappuccino", #Marco is a Analyst I at KPH and loves cappuccinos. 
    "Leah": "Pink Drink" #Leah is a Senior Analyst at KPH and loves pink drinks.
}

# We now determine what drink to order based on the drinks and the prices. 
for customer, drink in customers.items(): 
    if drink in drinks:
        # Indexing: place of an object in array: 
        # drinks = [0,1,2]
        # drinls [-2,-1,0] 
        # drinks = ["mocha", "latte", "frapuccino"]
        # drinks[1]? 
        # drinks[-1]? # last element? 
        # Same element! 
        
        # Slicing: drinks[0:2] = everything! 
        # Slicing question: drink[:-1]
        price = drinks[drink]
        # Python f string format a fancy way to print out results in a readable way. 
        print(f"{customer} ordered a {drink} for ${price:.2f}.")
    else:
        print(f"{customer} ordered a {drink}, but it is not on the menu.") 
        
# Return now the total cost of the drinks ordered. 
# List comprehension: "shortened way" to write a for loop that creates a new list based on an existing list. 

# List Comprehension: [expression for item in iterable if condition] [fancy for loop]

# Perfect Squares: [num for num in range(1, 11) if num**2 < 50] # [1, 2, 3, 4, 5, 6, 7] 
# Would 8 be an appropriate input for num? No: because 64 > 50. 

total_cost = sum(drinks[drink] for drink in customers.values() if drink in drinks) 
print(f"The total cost of the drinks ordered is ${total_cost:.2f}.")

# Function to determine the average drinks ordered. 
def determine_average_drinks_ordered(customers, drinks):
    total_cost = 0 # accumulator (add or subtract from a total)
    # .items() in Python represents a method that returns a view object that displays a list of a dictionary's key-value tuple pairs. 
    # We use an underscore (_) for the customer variable because we don't actually need it in this calculation. 
    
    # Dictionary: [key, value] = [_, drink] = [customer, drink] 
    for _, drink in customers.items(): 
        if drink in drinks:
            total_cost += drinks[drink] 
    average_cost = total_cost / len(customers)
    return average_cost 
```

```python
# Determine average cost of drinks ordered.
average_cost = determine_average_drinks_ordered(customers, drinks) 
print(f"The average cost of the drinks ordered is ${average_cost:.2f}.")
```
